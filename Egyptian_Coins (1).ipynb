```python
%%writefile streamlit_app.py
import streamlit as st
from ultralytics import YOLO
from collections import Counter
import cv2
import numpy as np
from PIL import Image




ONNX_MODEL_PATH = '/content/drive/MyDrive/YOLO_Coin_Detection_Results/Exp1_50e_Nano2/weights/best.onnx'


COIN_VALUES = {
    'gneh': 1.0,
    'nos_gneh': 0.5
}
CLASS_NAMES = ['gneh', 'nos_gneh']



@st.cache_resource
def load_yolo_model():
    """Loads the best ONNX model for inference."""
    try:

        model = YOLO(ONNX_MODEL_PATH)
        return model
    except Exception as e:
        st.error(f"Error loading model from Drive: {e}")
        st.stop()
        return None



def process_and_calculate_coins(model, source_image_bytes, confidence):
    """Runs inference and calculates the total coin value from an image."""


    image = np.array(Image.open(source_image_bytes).convert('RGB'))
    image_bgr = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)


    results = model(image_bgr, conf=confidence, iou=0.5, imgsz=640, verbose=False)


    result = results[0]


    detected_classes = [
        CLASS_NAMES[int(cls_id)]
        for cls_id in result.boxes.cls.tolist()
    ]


    coin_counts = Counter(detected_classes)


    total_value = 0.0
    for name, count in coin_counts.items():
        if name in COIN_VALUES:
            total_value += COIN_VALUES[name] * count


    annotated_image = result.plot()

    return annotated_image, coin_counts, total_value



st.set_page_config(
    page_title="Egyptian Coin Counter",
    layout="wide",
    initial_sidebar_state="expanded"
)

st.title("🪙 Egyptian Coin Counter & Detector")
st.markdown("---")


model = load_yolo_model()


st.sidebar.header("Model Configuration")
confidence = st.sidebar.slider("Confidence Threshold", 0.0, 1.0, 0.40)
st.sidebar.caption(f"Loaded Model: {ONNX_MODEL_PATH.split('/')[-1]}")


tab1, tab2 = st.tabs(["🖼️ Image Upload", "📸 Mobile Camera (Snapshot)"])



with tab1:
    st.header("Upload an Image")
    uploaded_file = st.file_uploader("Choose an image containing coins:", type=['jpg', 'jpeg', 'png'])

    if uploaded_file is not None:
        st.image(uploaded_file, caption='Uploaded Image', use_column_width=True)

        if st.button('Detect and Calculate Amount (Image)'):
            col1, col2 = st.columns(2)
            with st.spinner('Running detection...'):
                annotated_image, coin_counts, total_value = process_and_calculate_coins(model, uploaded_file, confidence)


                with col1:
                    st.image(cv2.cvtColor(annotated_image, cv2.COLOR_BGR2RGB), caption='Detected Coins', use_column_width=True)

                with col2:
                    st.subheader("Coin Calculation")
                    st.success(f"💰 **Total Value Detected: {total_value:.1f} EGP**")
                    st.markdown("---")
                    st.markdown("### Coin Breakdown")

                    for coin_name, count in coin_counts.items():
                        st.write(f"- **{coin_name.upper()}**: {count} coins x {COIN_VALUES[coin_name]:.1f} EGP = **{COIN_VALUES[coin_name] * count:.1f} EGP**")


with tab2:
    st.header("Live Camera Snapshot")
    st.warning("This mode captures a single frame from your device's camera for detection. **Access this app via the public URL on your mobile phone!**")


    camera_file_buffer = st.camera_input("Take a picture of the coins")

    if camera_file_buffer is not None:
        if st.button('Detect and Calculate Amount (Camera)'):
            col1, col2 = st.columns(2)
            with st.spinner('Running detection...'):

                annotated_image, coin_counts, total_value = process_and_calculate_coins(model, camera_file_buffer, confidence)


                with col1:
                    st.image(cv2.cvtColor(annotated_image, cv2.COLOR_BGR2RGB), caption='Detected Coins', use_column_width=True)

                with col2:
                    st.subheader("Coin Calculation")
                    st.success(f"💰 **Total Value Detected: {total_value:.1f} EGP**")
                    st.markdown("---")
                    st.markdown("### Coin Breakdown")

                    for coin_name, count in coin_counts.items():
                        st.write(f"- **{coin_name.upper()}**: {count} coins x {COIN_VALUES[coin_name]:.1f} EGP = **{COIN_VALUES[coin_name] * count:.1f} EGP**")
```
